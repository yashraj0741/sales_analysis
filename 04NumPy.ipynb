## 1. What is NumPy? (5 min)

**NumPy** means *Numerical Python*. Its central data structure is the **`ndarray`** (N-dimensional array), designed to store and operate on numerical data efficiently.

- A Python list is a general-purpose container.
- A NumPy array is designed for numerical computation and supports element-wise operations.
- **Vectorization** means applying an operation to an entire array without writing a Python loop.
- NumPy is often used alongside Pandas: Pandas organizes labelled/tabular data; NumPy powers many numerical operations beneath the surface.

### Excel-to-NumPy mental model
| Familiar idea | NumPy idea |
|---|---|
| A range of numeric cells | Array |
| Number format / value type | `dtype` |
| Rows and columns | `shape` |
| Formula filled down a column | Vectorized operation |
| SUM / AVERAGE / MIN / MAX | `np.sum`, `np.mean`, `np.min`, `np.max` |
| IF formula | `np.where` |

```python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

## 4. NumPy vs Pandas — Why Do We Need Both?

Pandas and NumPy both support mathematical operations, but they serve different purposes.

- **Pandas:** Works with labelled, tabular data. It is useful for analysing employee records, selecting columns, grouping departments, cleaning missing values and creating reports.
- **NumPy:** Works primarily with numerical arrays. It is useful for vectorized calculations, multidimensional arrays, broadcasting and mathematical operations.

### Example business problem

Our employee dataset contains a `Monthly_Salary` column. We want to calculate annual salaries and find the average salary.

We can solve this using either library.

**Key learning:** Pandas helps us manage and analyse the employee table; NumPy helps us perform numerical computation on arrays. They complement each other and are often used together.

```python
# Approach 1: Using Pandas
employee_df["Annual_Salary_Pandas"] = (
    employee_df["Monthly_Salary"] * 12
)

pandas_average = employee_df["Monthly_Salary"].mean()

# Approach 2: Using NumPy
salary_array = employee_df["Monthly_Salary"].to_numpy()

annual_salary_numpy = salary_array * 12
numpy_average = np.mean(salary_array)

print("Pandas average salary:", pandas_average)
print("NumPy average salary:", numpy_average)

print("\nAnnual salaries using NumPy:")
print(annual_salary_numpy)

display(employee_df[
    ["Employee_Name", "Monthly_Salary", "Annual_Salary_Pandas"]
])
```

## 5. Convert columns to arrays

Use `.to_numpy()` to get the underlying values from a Pandas Series. Keep employee names in the DataFrame; use numeric arrays for numerical calculations.

```python
salary = employee_df["Monthly_Salary"].to_numpy()
employee_ids = employee_df["Employee_ID"].to_numpy()

print("Salary array:", salary)
print("Salary shape:", salary.shape)
print("Salary dtype:", salary.dtype)
```

## 6. Indexing and slicing

- `salary[0]` selects the first value (Python uses **zero-based indexing**).
- `salary[-1]` selects the last value.
- `salary[1:4]` selects positions 1, 2, and 3; the stop position is excluded.
- For a 2-D array, use `array[row, column]`.

```python
print("First salary:", salary[0])
print("Last salary:", salary[-1])
print("Second to fourth salaries:", salary[1:4])

salary_matrix = np.array([[50000, 45000, 60000],
                          [72000, 68000, 55000]])
print("Matrix shape:", salary_matrix.shape)
print("Row 1, column 2:", salary_matrix[0, 1])
```

## 7. Vectorized arithmetic

A vectorized operation applies the same calculation to each value.

**Why it matters:** instead of looping through each employee to calculate annual salary, multiply the whole salary array by 12.

Be careful with units: monthly salary × 12 gives annual salary.

```python
annual_salary = salary * 12
salary_after_8_percent_raise = salary * 1.08
salary_in_thousands = salary / 1000

print("Annual salary:", annual_salary)
print("Salary after 8% raise:", salary_after_8_percent_raise)
print("Salary in thousands:", salary_in_thousands)
```

## 8. Aggregation and axis

Common summary functions:
- `np.sum()` — total
- `np.mean()` — average
- `np.median()` — median
- `np.min()` / `np.max()` — smallest / largest
- `np.std()` — standard deviation

For a 2-D array, `axis=0` works down rows (column-wise); `axis=1` works across columns (row-wise).

```python
print("Total monthly payroll:", np.sum(salary))
print("Average monthly salary:", np.mean(salary))
print("Median monthly salary:", np.median(salary))
print("Minimum salary:", np.min(salary))
print("Maximum salary:", np.max(salary))

quarterly_sales = np.array([[100, 120, 130],
                            [90, 110, 125]])
print("Total per quarter/column (axis=0):", np.sum(quarterly_sales, axis=0))
print("Total per employee/row (axis=1):", np.sum(quarterly_sales, axis=1))
```

## 9. Boolean masks and filtering 

A comparison such as `salary > 60000` creates a Boolean array (`True` / `False`). Use that mask to select values.

Use `&` for AND and `|` for OR, and put each condition in parentheses.

```python
high_salary_mask = salary > 60000
print("Mask:", high_salary_mask)
print("Salaries above 60000:", salary[high_salary_mask])

# Employees earning 50,000–70,000 inclusive
mid_salary_mask = (salary >= 50000) & (salary <= 70000)
display(employee_df.loc[mid_salary_mask, ["Employee_Name", "Monthly_Salary"]])
```

## 10. Business rules with `np.where()` and `np.select()`

- `np.where(condition, value_if_true, value_if_false)` is useful for a two-way rule (like Excel `IF`).
- `np.select(conditions, choices, default=...)` handles multiple rules, similar to nested `IF`.
- In `np.select`, condition order matters: put the highest threshold first.

```python
employee_df["Salary_Flag"] = np.where(salary >= 70000, "High", "Standard")

conditions = [
    salary >= 80000,
    salary >= 60000,
    salary >= 45000,
    salary < 45000
]
choices = ["Very High", "High", "Medium", "Low"]
employee_df["Salary_Band"] = np.select(conditions, choices, default="Unknown")

display(employee_df[["Employee_Name", "Monthly_Salary", "Salary_Flag", "Salary_Band"]])
```
