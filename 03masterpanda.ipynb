# Phase 10 — Date Handling

## 38. Extract date parts

Equivalent to Excel `YEAR()`, `MONTH()`, `DAY()`.

## 39. Calculate duration

Example: tenure from joining date.

## 40. Calculate tenure in years

Useful for HR analytics.

```python
# 37. Convert Join_Date to datetime if not already done
df_clean["Join_Date"] = pd.to_datetime(df_clean["Join_Date"])

# 38. Extract date parts
df_clean["Join_Year"] = df_clean["Join_Date"].dt.year
df_clean["Join_Month"] = df_clean["Join_Date"].dt.month
df_clean["Join_Day"] = df_clean["Join_Date"].dt.day

# 39 and 40. Calculate tenure
reference_date = pd.Timestamp.today().normalize()
df_clean["Tenure_Days"] = (reference_date - df_clean["Join_Date"]).dt.days
df_clean["Tenure_Years"] = (df_clean["Tenure_Days"] / 365).round(1)

display(df_clean[["Employee_Name", "Join_Date", "Join_Year", "Tenure_Years"]].head())
```

# Phase 16 — Apply, Map, Lambda

## 55. Apply function to a column

Use when logic is more complex than a simple formula.

## 56. Lambda function

Use for short one-line functions.

## 57. Apply across rows

Use `axis=1` when logic depends on multiple columns in the same row.

```python
# 55. Apply a named function to a column
def salary_category(salary):
    if salary >= 80000:
        return "High"
    elif salary >= 50000:
        return "Medium"
    else:
        return "Low"

df_clean["Salary_Band"] = df_clean["Monthly_Salary"].apply(salary_category)

# 56. Lambda function for annual salary
df_clean["Annual_Salary"] = df_clean["Monthly_Salary"].apply(lambda x: x * 12)

# Prepare attendance summary per employee
attendance_summary = attendance_df.groupby("Employee_ID").agg(
    Total_Hours=("Hours_Worked", "sum"),
    Total_OT_Hours=("Overtime_Hours", "sum"),
    Total_Leaves=("Leaves_Taken", "sum")
).reset_index()

payroll_df = df_clean.merge(attendance_summary, on="Employee_ID", how="left").merge(rate_df, on="Grade", how="left")

# 57. Apply across rows: total pay using rate card and overtime multiplier
def calculate_total_pay(row):
    regular_pay = row["Total_Hours"] * row["Base_Rate_Per_Hour"]
    overtime_pay = row["Total_OT_Hours"] * row["Base_Rate_Per_Hour"] * row["OT_Multiplier"]
    return regular_pay + overtime_pay

payroll_df["Calculated_Pay"] = payroll_df.apply(calculate_total_pay, axis=1)

display(payroll_df[["Employee_Name", "Grade", "Total_Hours", "Total_OT_Hours", "Calculated_Pay"]].head())
```

# Final Classroom Recap

Students who know Excel should now be able to map familiar Excel tasks to Pandas:

- Open workbook → `read_excel()`
- Open CSV → `read_csv()`
- Filter → Boolean conditions
- Sort → `sort_values()`
- Formula columns → new DataFrame columns
- IF / nested IF → `np.where()` / `np.select()`
- Remove duplicates → `drop_duplicates()`
- Text functions → `.str` methods
- Date formulas → `.dt` methods
- SUMIF / COUNTIF / AVERAGEIF → filtering + aggregation
- Pivot Table → `groupby()` / `pivot_table()`
- VLOOKUP / XLOOKUP → `map()` / `merge()`
- Append sheets/files → `concat()`
- Custom formulas → `apply()` and `lambda`
- Export report → `to_excel()` / `to_csv()`

This prepares students to move next into **visualisation and plotting in Pandas / Matplotlib / Seaborn**.
