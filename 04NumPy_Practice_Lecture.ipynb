## 4. Move a column into NumPy

Pandas is useful for working with named columns and records. With `.to_numpy()`, we can take a numeric column and perform array calculations on it.

```python
salary = employee_df["Monthly_Salary"].to_numpy()
employee_ids = employee_df["Employee_ID"].to_numpy()

print("Salary values:", salary)
print("Shape:", salary.shape)
print("Data type:", salary.dtype)
```

## 5. Indexing and slicing

Python starts counting positions at zero:
- `salary[0]` is the first salary.
- `salary[-1]` is the last salary.
- `salary[1:4]` returns positions 1, 2 and 3. The stop position is not included.

```python
print("First salary:", salary[0])
print("Last salary:", salary[-1])
print("Second to fourth salaries:", salary[1:4])
```

For a 2-D array, use `[row, column]`.

```python
salary_matrix = np.array([
    [50000, 45000, 60000],
    [72000, 68000, 55000]
])

print(salary_matrix)
print("Shape:", salary_matrix.shape)
print("Row 1, column 2:", salary_matrix[0, 1])
```

## 6. Calculate for every employee at once

NumPy lets us apply the same calculation to all values without writing a loop. This is called a *vectorized operation*.

```python
annual_salary = salary * 12
salary_after_raise = salary * 1.08
salary_in_thousands = salary / 1000

print("Annual salary:", annual_salary)
print("Salary after an 8% raise:", salary_after_raise)
print("Monthly salary in thousands:", salary_in_thousands)
```

## 7. Summarise the salary data

These functions are similar to common spreadsheet calculations:
- `np.sum()` adds the values
- `np.mean()` calculates the average
- `np.median()` finds the middle value
- `np.min()` and `np.max()` return the smallest and largest values
- `np.std()` measures how spread out the values are

```python
print("Total monthly payroll:", np.sum(salary))
print("Average monthly salary:", np.mean(salary))
print("Median monthly salary:", np.median(salary))
print("Lowest salary:", np.min(salary))
print("Highest salary:", np.max(salary))
print("Standard deviation:", np.std(salary))
```

For a 2-D array, `axis=0` adds down each column and `axis=1` adds across each row.

```python
quarterly_sales = np.array([
    [100, 120, 130],
    [90, 110, 125]
])

print("Column totals:", np.sum(quarterly_sales, axis=0))
print("Row totals:", np.sum(quarterly_sales, axis=1))
```

## 8. Filter with conditions

A comparison such as `salary > 60000` produces `True` or `False` for each value. Put that condition inside the array to keep only the matching values.

Use `&` for AND and `|` for OR. Put each condition in parentheses when combining them.

```python
above_60000 = salary[salary > 60000]
print("Salaries above 60000:", above_60000)

between_50000_and_70000 = salary[(salary >= 50000) & (salary <= 70000)]
print("Salaries from 50000 to 70000:", between_50000_and_70000)
```

To see the employees behind those values, apply the same condition to the DataFrame:

```python
mask = employee_df["Monthly_Salary"].to_numpy() > 60000
display(employee_df.loc[mask, ["Employee_Name", "Department", "Monthly_Salary"]])
```

## 9. Apply a simple business rule

`np.where()` works like a simple IF statement: if the condition is true, use one value; otherwise, use another.

```python
employee_df["Salary_Flag"] = np.where(salary >= 70000, "High", "Standard")
display(employee_df[["Employee_Name", "Monthly_Salary", "Salary_Flag"]])
```

For more than two labels, `np.select()` can be useful. Put the highest threshold first so a high salary isn’t caught by a lower threshold first.

```python
conditions = [
    salary >= 80000,
    salary >= 60000,
    salary >= 45000,
    salary < 45000
]
labels = ["Very High", "High", "Medium", "Low"]

employee_df["Salary_Band"] = np.select(conditions, labels, default="Check")
display(employee_df[["Employee_Name", "Monthly_Salary", "Salary_Band"]])
```
