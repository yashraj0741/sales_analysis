- `+` addition
- `-` subtraction

# 4. Variables

```python
student_name = "Asha"
student_age = 21

# Invalid examples:
# 2name = "Asha"
# student name = "Asha"
# class = "Python" 
```

# Variable Reassignment

```python
x = 10
print(x, type(x))

x = "10"
print(x, type(x))
```

# 7. Strings

- A string is a sequence of characters inside quotes.
- Single and double quotes can both be used.
- Strings support indexing and slicing.

```python
text = "Python"

print(text)
print(text[0])
print(text[-1])
print(text[0:3])
print(len(text))
```

# String Concatenation & f-Strings

- `+` joins strings.
- f-strings are usually clearer when inserting variables.

```python
first = "Python"
level = "Beginner"

print(first + " " + level)
print(f"{first} {level}")
```

# String Methods

- Methods are called with `.` after an object.
- Common examples: `.upper()`, `.lower()`, `.title()`, `.count()`, `.replace()`

```python
text = "hello world"

print(text.upper())
print(text.lower())
print(text.title())
print(text.count("l"))
print(text.replace("world", "Python"))
```

# 8. Why Collections?

- A single variable can store one value.
- Collections help organize multiple related values.
- Today: List → Dictionary → Tuple.

# 9. Lists

- Ordered collection.
- Mutable: elements can be changed.
- Created with square brackets `[]`.
- Can contain different data types and duplicates.

```python
countries = ["USA", "India", "China", "Brazil"]

print(countries)
print(type(countries))
print(len(countries))
```

# List Indexing

- Python uses zero-based indexing.
- Positive indexes start at `0`.
- Negative indexes start at `-1` from the end.

```python
countries = ["USA", "India", "China", "Brazil"]

print(countries[0])
print(countries[1])
print(countries[-1])
```

# List Slicing

- Syntax: `list[start:stop:step]`.
- The stop index is excluded.
- Omitted start/stop means use the beginning/end.

```python
countries = ["USA", "India", "China", "Brazil"]

print(countries[0:3])
print(countries[1:])
print(countries[:2])
print(countries[::2])
```

# List Membership

- Use `in` to check whether an element exists.
- `not in` checks the opposite.

```python
countries = ["USA", "India", "China", "Brazil"]

print("India" in countries)
print("Japan" in countries)
print("Japan" not in countries)
```

# List Mutation — The Key Idea

- Mutable means the existing list can be changed.
- Use indexing to update an element.

```python
numbers = [10, 20, 30,"rt"]

print(numbers)
numbers[0] = 100
print(numbers)
```

# Adding to Lists

- `append(x)` adds one item at the end.
- `insert(i, x)` adds at a position.
- `extend(iterable)` adds multiple elements.

```python
numbers = [1, 2, 3]

numbers.append(4)
print(numbers)

numbers.insert(1, 10)
print(numbers)

numbers.extend([20, 30])
print(numbers)
```

# append() vs extend()

- This distinction prevents a common beginner mistake.

```python
a = [1, 2, 3]

a.append([4, 5])
print(a)       # [1, 2, 3, [4, 5]]

b = [1, 2, 3]
b.extend([4, 5])
print(b)       # [1, 2, 3, 4, 5]
```

# Removing from Lists

- `remove(value)` removes the first matching value.
- `pop(index)` removes and returns an item.
- `del list[index]` deletes by index.
- `clear()` removes all elements.

```python
items = ["A", "B", "C", "D"]

items.remove("B")
print(items)

removed = items.pop(0)
print(removed, items)

del items[0]
print(items)

items.clear()
print(items)
```

# Sorting Lists

- `.sort()` changes the existing list.
- Use `reverse=True` for descending order.
- Later: `sorted()` returns a sorted result without changing the original.

```python
numbers = [4, 3, 10, 7, 1, 2]

numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)
```

# Copying Lists

- Use `copy()` or slicing `[:]` for a basic list copy.
- Avoid assuming `new = old` creates an independent list.

```python
countries = ["India", "China", "Brazil"]

a = countries.copy()
b = countries[:]

a[0] = "Japan"

print(countries)
print(a)
print(b)
```

# 10. Dictionaries

- Store data as `key: value` pairs.
- Dictionaries are mutable.
- Access values by key rather than numeric index.
- Keys must be unique.

```python
student = {
    "name": "Yash",
    "age": 25,
    "course": "Data Science"
}

print(student)
print(student["name"])
```

# Dictionary Mental Model

- Think of a dictionary as a labeled box: the key is the label, the value is the stored information.

```python
student = {
    "name": "Yash",
    "age": 25,
    "city": "Solapur"
}

# key  -> value
print(student["city"])
```

# Dictionary Add & Update

- Assigning a new key adds an item.
- Assigning an existing key updates its value.
- `.update()` can add/update multiple pairs.

```python
student = {"name": "Yash", "age": 25}

student["city"] = "Solapur"
student["age"] = 26
student.update({"course": "Python", "level": "Beginner"})

print(student)
```

# Dictionary: [] vs get()

- `dict[key]` raises `KeyError` when the key does not exist.
- `.get(key)` safely returns `None` (or a default) when absent.

```python
student = {"name": "Yash", "age": 25}

print(student["name"])
print(student.get("salary"))
print(student.get("salary", 0))
```

# Dictionary keys(), values(), items()

- `.keys()` → keys
- `.values()` → values
- `.items()` → key-value pairs

```python
student = {"name": "Yash", "age": 25}

print(student.keys())
print(student.values())
print(student.items())
```

# Dictionary Loop

- `.items()` is useful when you need both key and value.

```python
student = {"name": "Yash", "age": 25, "course": "Python"}

for key, value in student.items():
    print(key, ":", value)
```

# Dictionary Removal

- `pop(key)` removes and returns a value.
- `del dict[key]` deletes a key-value pair.
- `clear()` empties the dictionary.

```python
student = {"name": "Yash", "age": 25, "course": "Python"}

removed = student.pop("age")
print(removed, student)

del student["course"]
print(student)

student.clear()
print(student)
```

# 11. Tuples

- Ordered collection.
- Immutable: cannot be changed after creation.
- Can contain mixed data types.
- Created commonly with parentheses and commas.

```python
student = ("Yash", 25, "Python")

print(student)
print(type(student))
print(student[0])
print(student[-1])
```

# List vs Tuple

- List → mutable → `[ ]`
- Tuple → immutable → `( )`
- Both support ordering, indexing and slicing.

```python
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

my_list[0] = 100
print(my_list)

# my_tuple[0] = 100   # TypeError
```

# Tuple: The Single-Element Trap

- The comma is what makes a one-element tuple.
- `(10)` is an integer; `(10,)` is a tuple.

```python
a = (10)
b = (10,)

print(type(a))
print(type(b))
```

# Tuple Indexing, Slicing & Unpacking

- Tuples support indexing and slicing.
- Unpacking assigns tuple elements to variables.

```python
student = ("Yash", 25, "Python")

print(student[0])
print(student[1:])
print(student[:2])

name, age, course = student
print(name)
print(age)
print(course)
```

# Tuple Concatenation

- Use `+` to combine tuples.
- The result is a new tuple.

```python
a = (1, 2)
b = (3, 4)

c = a + b
print(c)
```

# 12. Nested Collections

- Collections can contain other collections.
- This is common in real datasets.

```python
students = [
    {"name": "Asha", "marks": 85},
    {"name": "Rahul", "marks": 92}
]

print(students[0]["name"])
print(students[1]["marks"])
```

# 13. One End-to-End Example

- Use multiple concepts together.
- Ask students to predict the output before running it.

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

skills = ["Python", "SQL", "Power BI"]

student = {
    "name": name,
    "age": age,
    "skills": skills
}

print(f"{student['name']} is {student['age']} years old.")
print("Skills:", student["skills"])
print("Number of skills:", len(student["skills"]))
```
