## Phase 1 — Why plotting is needed in data analysis

Tables and summary statistics are useful, but they do not always reveal patterns clearly.

A plot can quickly show:

1. **Trend** — how something moves over time  
2. **Comparison** — which department, product, employee, or category is higher/lower  
3. **Distribution** — how values are spread  
4. **Relationship** — whether two variables move together  
5. **Outliers** — unusually high or low values  
6. **Composition** — how a total is divided into parts  
7. **Ranking** — top and bottom performers

In Excel, students may use chart menus. In Python, we first prepare the DataFrame and then call the right plotting command.

### 3.4.2 Additional finer control using matplotlib
A plot has several parts. Students should learn these names early because these terms appear repeatedly in Matplotlib and Seaborn.

1. **Figure** — the full chart area  
2. **Axes** — the actual plotting area  
3. **X-axis** — horizontal axis
4. **Y-axis** — vertical axis  
5. **Title** — what the chart is about  
6. **Labels** — explanation of each axis  
7. **Legend** — explanation of colors or groups  
8. **Grid** — reference lines  
9. **Ticks** — values shown on axes  
10. **Figure size** — width and height  
11. **Rotation of labels** — useful for long category names  
12. **Tight layout** — avoids cutting labels

```python
plt.plot(df_channel_spend['Year'],df_channel_spend['TV_Spend_USD'], label = 'TV', color ='blue')
plt.plot(df_channel_spend['Year'], df_channel_spend['Radio_Spend_USD'], label ='Radio', color = 'green')
plt.title ('channel spend  trend')
plt.xlabel('Year')
plt.ylabel('Spend')
plt.legend(title="Marketing Channels")
plt.grid(True)
plt.show()
```

### 3.5  Seaborn Plot for Statistical Plot

#### 3.5.1. Distribution Plot

```python
# ===============================
# DISTRIBUTION PLOT
# ===============================
plt.figure(figsize=(8,5))
sns.histplot(df_channel_spend['TV_Spend_USD'], kde=True) #kde is kernel density estimate -shows proability density spread
plt.title("Distribution of TV Spend with Density Curve")
plt.show()
```

### 3.5.2 Scatter + Regression Line

```python
# ===============================
#  SCATTER + REGRESSION LINE
# ===============================
plt.figure(figsize=(8,5))
sns.regplot(x='TV_Spend_USD', y='Revenue_USD', data=df_channel_spend)
plt.title("TV Spend vs Revenue (With Regression Line)")
plt.show()
```

### 3.5.3 Correlation Heatmap

```python
df = df_campaign[['TV_Spend_USD', 'Radio_Spend_USD', 'Digital_Spend_USD', 'Revenue_USD']]
# ===============================
# CORRELATION HEATMAP
# ===============================
plt.figure(figsize=(8,6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

### 3.6 Plotly Interactive Plot

```python
fig = px.line(df_channel_spend, x='Year', y='Revenue_USD')
fig1= px.line(df_channel_spend, x='Year', y='TV_Spend_USD')
fig.show()
fig1.show()
```

## Conclusion — Plotting libraries: role clarity

There are four common ways to plot from a DataFrame.

1. **Pandas built-in plotting**  
   - Quick and convenient for early exploration  
   - Example: `df["Salary"].plot(kind="hist")`

2. **Matplotlib**  
   - Base plotting library  
   - Gives high control over titles, labels, grids, legends, figure size, and saving charts  
   - Common alias: `plt`

3. **Seaborn**  
   - Statistical visualization library built on Matplotlib  
   - Cleaner syntax for EDA plots like `histplot`, `boxplot`, `scatterplot`, `heatmap`, `barplot`, and `countplot`

4. **Plotly**  
   - Used for interactive plot.
   - When stakeholders need more clarity and they want graph to present exact data and not leave things for implicit interpretation


**Practical recommendation:** Use Pandas for quick plots, Matplotlib for control, and Seaborn for attractive statistical plots and plotly for interactive presentation.
